# Run your first Kotlin notebook cell

In this notebook, you will learn how to select the Kotlin kernel, run a code cell, and find its output.

## What is a Jupyter notebook?

A **Jupyter notebook** is a document that combines explanations with code you can run. A **text cell**, like this one, contains instructions. A **code cell** contains instructions for the computer.

In this class, you will use notebooks in **Grader Than Workspace** to read explanations, watch demonstrations, run examples, and practice writing code. You can work through a lesson and try its examples in the same document.

A **kernel** is the program that runs your code. Select the kernel for the language used in this notebook: **Kotlin**. The **output** is the result shown beneath a cell after you run it. Simply reading or editing a cell does not run it.

## Try it yourself

The code below displays a short greeting. You do not need to understand its punctuation yet; focus on selecting the kernel and running the cell.

1. Select the kernel control at the top right of the notebook. It may say **Select Kernel** or show a kernel name. Select this control even if **Java** is already selected.
2. Choose **Jupyter Kernel...**, then **Kotlin**. Wait for the kernel to finish starting.
3. Move to the code cell below and select its triangular **Run** button on the left. Use the cell button shown in the demonstration.
4. Look directly below the cell for its output.

````kotlin
println("Hello, world!")
````

## Check the result

After the cell runs, you should see this output directly beneath it:

```text
Hello, world!
```

If you see that greeting without an error, you have successfully run the cell. You can run this greeting cell again to display the same message.

## If the cell does not run

Give the kernel a moment to start. If the notebook asks you to select a kernel, follow the steps above and choose the Kotlin kernel. If that kernel is missing or an error remains, ask your instructor for help. You do not need to install anything to complete this activity.
