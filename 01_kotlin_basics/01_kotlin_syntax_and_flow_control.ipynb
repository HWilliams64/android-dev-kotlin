## Why This Matters

An Android app repeatedly turns data into decisions: whether a button is available, which message a user sees, or how many items appear in a list. Kotlin's expression syntax lets you keep a selected value close to the condition that determines it. Choosing a non-reassignable name also makes it clearer which values should stay fixed while an operation runs.

You could express these rules in Java or C++. Learning Kotlin's conventions helps you read Android examples accurately and avoid carrying over rules that differ, such as Java's use of `==` for object references. We will reuse this foundation when functions, missing data, and app objects enter the picture.

## Concept

A student revises a note called **Lab ideas**. The app holds a title and a revision count. It needs to tell the student whether a title is present and whether the note is new, revised once, or revised several times. We will also print two numbered previews. A revision count is a whole-number count, not a time or a file size.



### Names, types, and reassignment

A **binding** associates a name with a value. Kotlin's **`val`** keyword creates a binding that you cannot reassign. Its **`var`** keyword allows reassignment. Prefer `val` when the operation does not need to replace the value.

```kotlin
val noteTitle = "Lab ideas"
var revisionCount = 1
revisionCount += 1
```

The first line gives the text a name. The second starts a count at one. The third increases that count by one, leaving it at two. `+=` is an assignment operator. It is not a new variable declaration. Kotlin usually needs neither a semicolon at the end of a line nor Java's `new` syntax; object construction comes later.

**Type inference** means the compiler determines a type from the assigned expression. `noteTitle` is a `String`, and `revisionCount` is an `Int`. Inference does not make Kotlin dynamically typed: assigning text to this integer variable would still be an error. You may state the type after a colon when it improves clarity.

```kotlin
val maximumPreviews: Int = 2
val isSaved: Boolean = true
val syncProgress: Double = 0.5
```

`Int` represents a whole number, `Boolean` represents true or false, and `Double` represents a floating-point number. They are type names, not variable-declaration keywords. Kotlin also has `Long`, `Float`, and `Char`; we will introduce extra types only when an example needs them. Numeric assignment does not silently convert an `Int` into a `Double`; use an appropriate literal such as `2.0` for a `Double`.

A `val` prevents replacing the value through that name. It does **not** promise that an object reached through the name can never change. We will examine that distinction with collections.



The runnable cell below displays each value in declaration order. `println` is the standard-library function that displays a value followed by a new line. The output makes both the count update and the declared values visible.

The short demonstrations repeat their input declarations so you can run each one on its own. Re-evaluating a declaration in a notebook cell is different from assigning a new value to an existing `val` within one execution.

````kotlin
val noteTitle: String = "Lab ideas"
var revisionCount = 1
revisionCount += 1
val maximumPreviews: Int = 2
val isSaved: Boolean = true
val syncProgress: Double = 0.5
println(noteTitle)
println(revisionCount)
println(maximumPreviews)
println(isSaved)
println(syncProgress)
````

The count is `2` after the update. The other outputs show the title, the preview limit, the Boolean value, and the floating-point value. They remain separate values with separate types.

### Building text and comparing values

A **string template** inserts a value into text. Write `$name` for one variable or `${expression}` for an expression.

```kotlin
println("$noteTitle has $revisionCount revisions")
println("Next revision: ${revisionCount + 1}")
```

The first line inserts both current values. The second calculates a value for the message; it does not update `revisionCount`. `println` is a standard-library function that displays a value followed by a new line. It is not a keyword.

Kotlin's **structural equality** operator, `==`, compares values. For strings, it compares their text. Use `!=` to check that values differ. This contrasts with Java's reference comparison for objects using `==`. Kotlin also has reference equality, `===`, but our text checks need `==`.

```kotlin
val titleIsEmpty = noteTitle == ""
```

An empty string contains no characters. A string containing a space is not empty. This lesson checks only empty text; it does not yet trim or normalize a user's input.



````kotlin
val noteTitle = "Lab ideas"
val revisionCount = 2
println("$noteTitle has $revisionCount revisions")
println("Next revision: ${revisionCount + 1}")
println(noteTitle == "Lab ideas")
println(noteTitle != "")
println(revisionCount)
````

Both comparisons produce `true`. The final `2` confirms that calculating the next revision inside a template left the stored count unchanged.

### Selecting one value with if

In Kotlin, **`if`** can be an **expression**: it can produce a value. That means a condition can directly determine what a variable receives.

```kotlin
val titleMessage = if (noteTitle == "") "Add a title" else "Title ready"
```

The condition checks whether the title is empty. If it is, the first text becomes the message. Otherwise, the **`else`** branch supplies the second text. Because this `if` must produce a value, both possibilities need a result. Kotlin does not use Java/C++'s `condition ? first : second` form for this choice. Larger branch bodies may use braces; their last expression supplies the branch's value.



````kotlin
val noteTitle = ""
val titleMessage = if (noteTitle == "") "Add a title" else "Title ready"
println(titleMessage)
````

This demonstration deliberately uses empty text, so the selected result is `Add a title`. Changing the input to a nonempty title selects the other branch.

### Selecting among several values with when

We now need three possible revision messages. The **`when`** keyword selects the first matching branch. Here, each arrow separates a count from its resulting message.

```kotlin
val revisionMessage = when (revisionCount) {
    0 -> "New note"
    1 -> "First revision"
    else -> "Multiple revisions"
}
```

A zero count chooses the first message. A count of one chooses the second. Other counts reach `else`. Our app's input rule is that revision counts are nonnegative; a negative count would also reach `else`, so this code alone does not validate that rule. There is no fall-through from one branch into the next and no `break` needed between these branches.

For a separate worked case, a count of **one** selects **First revision**. The zero branch does not match, the one branch does, and the selection ends. That reasoning also works for the count of two in our complete example.

````kotlin
val revisionCount = 1
val revisionMessage = when (revisionCount) {
    0 -> "New note"
    1 -> "First revision"
    else -> "Multiple revisions"
}
println(revisionMessage)
````

For this worked case, the output is `First revision`: the count is one and only that branch matches.

## Worked Example

### Repeat a preview with a range

A **range** describes a progression of values. With integers, `1..2` includes both one and two. The **`for`** keyword repeats its body for each value, and **`in`** connects the loop variable to the range.

```kotlin
for (previewNumber in 1..2) {
    println("Preview $previewNumber")
}
```

On the first pass, `previewNumber` is one. On the second, it is two. Each pass inserts the current number into the message. You do not declare the loop variable separately.

An ascending range such as `1..0` is empty. It runs the body zero times; it does not count backward. This is useful when a preview count is zero. The range `0..<2` includes zero and one while excluding two. The operators `..` and `..<` are not keywords. We use the inclusive form below because our visible numbering begins at one.



````kotlin
for (previewNumber in 1..2) {
    println("Preview $previewNumber")
}
for (indexNumber in 0..<2) {
    println("Index $indexNumber")
}
for (emptyNumber in 1..0) {
    println("Unexpected $emptyNumber")
}
println("Ranges finished")
````

The first loop prints previews one and two. The next prints index labels zero and one. No `Unexpected` line appears because `1..0` is empty. `Ranges finished` confirms that execution continued after that empty loop.

### Repeat while a condition holds

Kotlin's **`while`** loop follows the condition-first pattern you know from Java and C++. Its condition is checked before every pass.

```kotlin
var remainingPreviews = 2
while (remainingPreviews > 0) {
    println("Remaining: $remainingPreviews")
    remainingPreviews -= 1
}
```

This prints two, then one, as part of the labeled messages. Reducing the count makes the condition false and stops the loop. Without the update, this example would keep repeating. Use `for` when a range gives the values you need, or `while` when a condition controls repetition.



````kotlin
var remainingPreviews = 2
while (remainingPreviews > 0) {
    println("Remaining: $remainingPreviews")
    remainingPreviews -= 1
}
````

The remaining count decreases from two to one to zero. At zero, the condition is false, so the loop stops before printing another line.

### Assemble the note status

The complete program below connects the pieces: store the inputs, record a revision, choose messages, then display two previews. Unlike a compiled Java application, a Kotlin notebook cell can evaluate these top-level statements directly. A `main` function is unnecessary here.

````kotlin
val noteTitle = "Lab ideas"
var revisionCount = 1
revisionCount += 1
val titleMessage = if (noteTitle == "") "Add a title" else "Title ready"
val revisionMessage = when (revisionCount) {
    0 -> "New note"
    1 -> "First revision"
    else -> "Multiple revisions"
}
println("$noteTitle: $titleMessage")
println(revisionMessage)
for (previewNumber in 1..2) {
    println("Preview $previewNumber")
}
````

The note has a nonempty title, so the first output line ends with **Title ready**. The count starts at one and becomes two, so `when` chooses **Multiple revisions**. The inclusive range then produces **Preview 1** and **Preview 2**.

Notice that calculating a message and displaying it are separate operations. The `if` and `when` expressions produce strings; `println` displays those strings. Keeping these responsibilities distinct will help when an Android screen eventually displays the same information.

## Summary

- Choose `val` for a binding you will not reassign and `var` when an update is needed. Inferred types remain fixed.
- Use string templates to build messages and `==` to compare string contents.
- `if` and `when` can produce a selected value. `when` runs only the first matching branch.
- `1..count` includes its upper bound; it is empty when `count` is zero. A `while` loop repeats while its condition remains true.

You can now express the same basic decisions and repetitions you already know using Kotlin syntax. In the next lesson, functions will package these rules for reuse, and null safety will help distinguish missing information from existing text.
